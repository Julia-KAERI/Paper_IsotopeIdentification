```python
import numpy as np
import matplotlib.pyplot as plt

def crs(d):
    xr = np.arange(-15, 15.05, 0.1)
    yr = np.arange(-90, 90.05, 0.1)
    dx = xr[1]-xr[0]
    dy = yr[1]-yr[0]
    X, Y = np.meshgrid(xr, yr)

    Rd, Rxy = np.sqrt(X**2+Y**2+d**2), np.sqrt(X**2+Y**2)
    return np.sum(dx*dy*Rxy/(Rd**3))
    
```

```python
crs(100)/crs(80)
```

```python
crs(80)
```

```python
dd = np.arange(60, 200, 5)
cc = [crs(d) for d in dd]
```

```python
plt.plot(dd, cc)
plt.plot(dd, 1800/dd**2)
```
