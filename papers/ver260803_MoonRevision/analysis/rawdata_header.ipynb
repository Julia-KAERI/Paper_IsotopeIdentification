```python
import glob, re, sys, os
sys.path.insert(1, os.path.realpath(os.path.pardir))
from specs import *
import pandas as pd
from scipy.optimize import curve_fit
from scipy.stats import t
from matplotlib.transforms import ScaledTranslation
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "Helvetica"
})

current_paper_directory = "../newfigures"
FLAG_savefig=False

rawDir = "../../../data/"
pn1Dir = rawDir + "P1_SN1222/"
pn2Dir = rawDir + "P2_SN1215/"
```

```python
rows = []
for path in sorted(glob.glob(os.path.join(pn1Dir,'*.mca'))):
    lines = [l.strip() for l in open(path, encoding='latin-1').read().splitlines()]
    def hdr(key):
        for l in lines:
            if l.startswith(key):
                return l.split('-', 1)[1].strip()
        return ''
    def status(key):
        for l in lines:
            if l.strip().startswith(key):
                return l.split(':', 1)[1].strip()
        return ''
    def cfg(key):
        for l in lines:
            if l.startswith(key + '='):
                return l.split('=', 1)[1].split(';')[0]
        return ''
    rows.append([
        os.path.basename(path),
        hdr('LIVE_TIME'), hdr('REAL_TIME'), hdr('START_TIME'),
        status('Fast Count'), status('Slow Count'), status('Dead Time'),
        cfg('MCAC'), cfg('GAIN'), cfg('THFA'), cfg('THSL'),
    ])

print('file\tlive\treal\tstart\tfast\tslow\tdead\tchan\tgain\tthfa\tthsl')
for r in rows:
    print('\t'.join(str(x) for x in r))
```
